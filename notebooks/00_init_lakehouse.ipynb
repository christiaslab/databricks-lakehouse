# Init Lakehouse
Creates the catalog objects the project needs: the three medallion schemas and the volume where raw CSV files are uploaded.

Run once, safe to re-run (everything is `IF NOT EXISTS`).

## Select catalog

```python
%sql
USE CATALOG workspace;
```

## Create medallion schemas

```python
%sql
CREATE SCHEMA IF NOT EXISTS bronze
COMMENT 'Bronze layer: raw ingested data';

CREATE SCHEMA IF NOT EXISTS silver
COMMENT 'Silver layer: cleaned and standardized data';

CREATE SCHEMA IF NOT EXISTS gold
COMMENT 'Gold layer: business-ready dimensional model';
```

## Create volume for raw source files
Upload the CSVs from `data/` into `src/src_crm/` and `src/src_erp/`.

```python
%sql
CREATE VOLUME IF NOT EXISTS workspace.bronze.src
COMMENT 'Raw source files (CSV) from CRM and ERP systems';
```
