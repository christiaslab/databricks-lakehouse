Runs every silver notebook in sequence. This is the single entry point for the silver layer when scheduling a job.

Each notebook is independent (reads bronze, writes its own silver table), so order does not matter — they run sequentially only to keep things simple.

```python
import os

# Resolve paths relative to this notebook, so it works from any workspace location (git folder, sync folder, job)
here = os.path.dirname(dbutils.notebook.entry_point.getDbutils().notebook().getContext().notebookPath().get())

NOTEBOOKS = [
    "crm/silver_crm_cust_info",
    "crm/silver_crm_prd_info",
    "crm/silver_crm_sales_details",
    "erp/silver_erp_cust_az12",
    "erp/silver_erp_loc_a101",
    "erp/silver_erp_px_cat_g1v2",
]

for nb in NOTEBOOKS:
    path = f"{here}/{nb}"
    print(f"Running {path}")
    dbutils.notebook.run(path, timeout_seconds=0)
    print("  done")
```
